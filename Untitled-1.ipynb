2. A Correlação Real entre Distância e Tempo (r = 0.704)

Antes da limpeza, a correlação estava artificialmente próxima de zero (0.032) devido à poluição dos outliers.

Após isolar os dados válidos, a matriz de correlação mostra um valor de 0.7038 entre distancia_km_limpa e tempo_entrega_limpo.

O que isto significa: Existe agora uma forte correlação linear positiva (cerca de 70%), o que faz todo o sentido do ponto de vista operacional — quanto maior é a distância percorrida pelo entregador, maior tende a ser o tempo necessário para a conclusão da entrega.

Deste modo, a base de dados está tecnicamente limpa e com padrões lógicos consistentes, pronta para avançar para as análises temporais e cruzamentos com dados climáticos (chuva) ou de faturamento.

```python
import unicodedata

def normalizar_texto(texto):
    if not isinstance(texto, str):
        return texto

    texto = texto.strip().lower()
    texto = unicodedata.normalize("NFKD", texto)
    texto = "".join(
        c for c in texto
        if not unicodedata.combining(c)
    )

    return texto
```

```python
import unicodedata

# 1. Função de normalização (caso ainda não esteja carregada no seu notebook)
def normalizar_texto(texto):
    if texto is None:
        return texto
    texto = str(texto).strip().lower()
    texto = unicodedata.normalize("NFKD", texto)
    texto = "".join(
        caractere for caractere in texto
        if not unicodedata.combining(caractere)
    )
    return texto

# 2. Criar a coluna 'cidade_chave' a partir da coluna 'cidade' original
df["cidade_chave"] = df["cidade"].apply(normalizar_texto)

# 3. Dicionário de mapeamento para os valores normalizados
mapa_cidades = {
    "sao paulo": "São Paulo",
    "sp": "São Paulo",
    "rio de janeiro": "Rio de Janeiro",
    "rj": "Rio de Janeiro",
    "rio": "Rio de Janeiro",
    "belo horizonte": "Belo Horizonte",
    "bh": "Belo Horizonte",
    "b. horizonte": "Belo Horizonte"
}

# 4. Criar a coluna limpa definitiva
df["cidade_limpa"] = df["cidade_chave"].map(mapa_cidades)

# 5. Conferir o resultado final
print(df["cidade_limpa"].value_counts(dropna=False))
```

```python
Q1 = df["valor_pedido_limpo"].quantile(0.25)
Q3 = df["valor_pedido_limpo"].quantile(0.75)

IQR = Q3 - Q1
limite_superior = Q3 + 1.5 * IQR

print("Limite superior:", limite_superior)

df[df["valor_pedido_limpo"] > limite_superior][
    ["pedido_id", "valor_pedido", "valor_pedido_limpo"]
]
```

```python
df_sem_outliers = df[
    df["valor_pedido_limpo"] <= limite_superior
]

print("Média original:",
      df["valor_pedido_limpo"].mean())

print("Média sem outliers:",
      df_sem_outliers["valor_pedido_limpo"].mean())

print("Mediana original:",
      df["valor_pedido_limpo"].median())

print("Quantidade de outliers:",
      (df["valor_pedido_limpo"] > limite_superior).sum())
```

Os valores extremos estão puxando a média para cima. Quando os 352 valores acima do limite são retirados, a média cai de R$ 71,55 para R$ 64,93.

A mediana de R$ 61,61 também é menor que a média original, outro sinal de que existem valores altos influenciando a distribuição.

```python
# Quartis
Q1 = df["valor_pedido_limpo"].quantile(0.25)
Q3 = df["valor_pedido_limpo"].quantile(0.75)

# Intervalo interquartil
IQR = Q3 - Q1

# Limites pelo critério de 1,5 x IQR
limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

print(f"Q1: R$ {Q1:.2f}")
print(f"Q3: R$ {Q3:.2f}")
print(f"IQR: R$ {IQR:.2f}")
print(f"Limite inferior: R$ {limite_inferior:.2f}")
print(f"Limite superior: R$ {limite_superior:.2f}")
```

```python
IQR = Q3 - Q1
# 85.12 - 45.22 = 39.90

limite_superior = Q3 + 1.5 * IQR
# 85.12 + (1.5 * 39.90) = 144.97
```

```python
outliers = df[
    (df["valor_pedido_limpo"] < limite_inferior) |
    (df["valor_pedido_limpo"] > limite_superior)
]

print(f"Total de registros: {len(df)}")
print(f"Quantidade de outliers: {len(outliers)}")
print(f"Percentual de outliers: {len(outliers) / len(df) * 100:.2f}%")
```

Pelo método do intervalo interquartil (IQR), foram identificados 352 valores atípicos na variável valor_pedido, correspondentes a 4,37% dos 8.058 registros. O limite superior calculado foi de R$ 144,97. Esses registros devem ser analisados com cautela, pois a classificação como outlier estatístico não significa necessariamente que sejam dados incorretos.
